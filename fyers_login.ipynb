# 🔑 Fyers Daily Login Generator
Run the cell below to generate your daily Fyers Access Token. This works directly in your web browser on both **Mobile** and **Desktop**!

```python
!pip install fyers-apiv3 -q
import urllib.parse
from fyers_apiv3 import fyersModel
from IPython.display import HTML, display

CLIENT_ID = input('Enter your Fyers App ID (e.g., ABCD123-100): ').strip()
SECRET_KEY = input('Enter your Secret Key: ').strip()
REDIRECT_URI = 'https://fyers-realtime-p3sdaomlya-el.a.run.app/api/auth/callback'

if not CLIENT_ID.endswith('-100'):
    CLIENT_ID = f'{CLIENT_ID}-100'

session = fyersModel.SessionModel(
    client_id=CLIENT_ID,
    secret_key=SECRET_KEY,
    redirect_uri=REDIRECT_URI,
    response_type='code',
    grant_type='authorization_code'
)

auth_url = session.generate_authcode()
print('\n' + '='*50)
print('1. CLICK THE LINK BELOW TO LOGIN TO FYERS:')
print('='*50)
display(HTML(f'<a href="{auth_url}" target="_blank" style="font-size: 16px; color: #007bff; font-weight: bold;">Click Here to Login</a>'))
print('\n(After logging in, you will be redirected to an Error page. Copy the URL from your browser address bar.)\n')

redirected_url = input('2. Paste the FULL redirected URL here: ').strip()

try:
    if 'auth_code=' in redirected_url:
        auth_code = redirected_url.split('auth_code=')[1].split('&')[0]
    elif redirected_url.startswith('eyJh'):
        auth_code = redirected_url.split('&')[0]
    else:
        print('\nError: Could not find auth_code.')
        auth_code = None
        
    if auth_code:
        session.set_token(auth_code)
        response = session.generate_token()
        
        if response.get('s') == 'ok':
            print('\n\n✅ SUCCESS! Copy this Access Token:\n')
            print(response['access_token'])
            print('\n→ Now go to GitHub > cpr-screener > Settings > Secrets and Variables > Actions')
            print('→ Update the FYERS_ACCESS_TOKEN secret with the text above.')
        else:
            print(f'\nError from Fyers: {response}')
except Exception as e:
    print(f'Error: {e}')
```
